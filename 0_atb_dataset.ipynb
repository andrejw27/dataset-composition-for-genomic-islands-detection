```python
import gzip
import csv
import subprocess

tsv_gz_file = f"{FOLDER_TO_ATB}/file_list.all.latest.tsv.gz"

tar_xz_files = {}

with gzip.open(tsv_gz_file, "rt") as file:
    reader = csv.DictReader(file, delimiter = "\t")
    for row in reader:
        tar_xz = row["tar_xz"]
        tar_xz_url = row["tar_xz_url"]
        if tar_xz not in tar_xz_files:
            tar_xz_files[tar_xz] = tar_xz_url
```

```python
# Downloading the .fa files

from tqdm import tqdm 

for tar_xz, url in tqdm(selected_files_dict.items()):
    
    dest_folder = f"{FOLDER_TO_ATB}/atb_batch"
    os.makedirs(dest_folder, exist_ok=True)

    dest_file = f"{dest_folder}/{tar_xz}"
    if os.path.exists(dest_file) == False:
        print(f"Downloading {tar_xz} from {url}")
        subprocess.run(["wget", "-O", dest_file, url])
```

```python
from ete3 import NCBITaxa

ncbi = NCBITaxa()

def species_to_lineage(species_name):
    try:
        # 1. Translate name to TaxID
        name_map = ncbi.get_name_translator([species_name])
        if not name_map:
            return "Name not found"
        taxid = name_map[species_name][0]
        
        # 2. Get the lineage (list of parent TaxIDs)
        lineage = ncbi.get_lineage(taxid)
        
        # 3. Get the rank for each ID in the lineage
        ranks = ncbi.get_rank(lineage)  # Returns {taxid: 'rank_name'}
        
        # 4. CORRECTED: Use get_taxid_translator to get names
        # Returns {taxid: 'name'}
        names = ncbi.get_taxid_translator(lineage)
        
        # 5. Build a clean dictionary of the desired ranks
        desired_ranks = ['phylum', 'class', 'order', 'family']
        taxonomy = {ranks[tid]: names[tid] for tid in lineage if ranks[tid] in desired_ranks}
        
        return taxonomy
    except Exception as e:
        return f"Error: {e}"

print(species_to_lineage("Bacillus"))
```

### Assign taxonomy to the samples

```python
from Bio import Entrez
import taxoniq

Entrez.email = "your_email@example.com"

def get_taxid_from_sample(biosample_id):
    # Search for the BioSample
    handle = Entrez.esearch(db="biosample", term=biosample_id)
    record = Entrez.read(handle)
    ids = record["IdList"]
    
    if not ids:
        return None
    
    # Fetch the details
    handle = Entrez.esummary(db="biosample", id=ids[0])
    summary = Entrez.read(handle)

    taxid = int(summary["DocumentSummarySet"]["DocumentSummary"][0].get("Taxonomy"))
    
    # Extract TaxID (usually stored in 'TaxId' or within the SampleData XML)
    # BioSample summaries are often returned as XML strings in 'SampleData'
    return taxid

# map the taxid to the samples
def map_taxon_to_df(row):
    try:
        taxon = taxoniq.Taxon(row['taxid'])
        row['scientific_name'] = taxon.scientific_name

        for t in taxon.ranked_lineage:
            row[t.rank.name] = t.scientific_name 
    except Exception as e:
        row['scientific_name'] = None
    return row

def map_biosample_to_lineage(biosample_id):
    taxid = get_taxid_from_sample(sample_id)
    info = map_taxon_to_df({'taxid': taxid})

    return {biosample_id: info}


# Example use:
sample_id = "SAMEA7290760"
#taxid = get_taxid_from_sample(sample_id)
#info = map_taxon_to_df({'taxid': taxid})

map_biosample_to_lineage(sample_id)
```
