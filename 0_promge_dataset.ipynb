```python
promge_df = pd.read_csv("dataset/prev_studies/proMGE/mges.txt", sep="\t")

promge_df['taxid'] = promge_df.apply(lambda x: int(x["mge_genome_position"].split(".")[0]), axis=1)
promge_df['biosample_id'] = promge_df.apply(lambda x: x["mge_genome_position"].split(".")[1], axis=1)
promge_df['position'] = promge_df.apply(lambda x: x["mge_genome_position"].split(".")[-1], axis=1)
promge_df['Accession'] = promge_df['position'].str.extract(r'^([^:]+)')
# Apply extract to create two new columns
promge_df[['Start', 'End']] = promge_df['position'].str.extract(r':(\d+)-(\d+)').astype(int)
```

## Read proMGE HGT data
- source: https://promge.embl.de/about.cgi 

```python
# read text file into pandas DataFrame
promge_hgt_df = pd.read_csv("dataset/prev_studies/proMGE/HGT_data.txt", sep="\t")
```

```python
promge_hgt_df.head()
```

```python
# compute unique transfer e.g. A<->B is counted once

# ensure consistent ordering within each pair
promge_hgt_df[['m1', 'm2']] = pd.DataFrame(
    promge_hgt_df.apply(lambda x: sorted([x['HGT_genome_1'], x['HGT_genome_2']]), axis=1).tolist(),
    index=promge_hgt_df.index
)

# drop duplicates
unique_df = promge_hgt_df.drop_duplicates(subset=['recombinase_cluster', 'm1', 'm2'])
unique_df['bio_sample_id_1'] = unique_df['HGT_genome_1'].apply(lambda x: x.split('.')[-1])
unique_df['bio_sample_id_2'] = unique_df['HGT_genome_2'].apply(lambda x: x.split('.')[-1])
unique_df['accession_1'] = unique_df['MGE_position_1'].apply(lambda x: x.split('.')[-1].split(':')[0])
unique_df['accession_2'] = unique_df['MGE_position_2'].apply(lambda x: x.split('.')[-1].split(':')[0])
unique_df['HGT_genus_1'] = unique_df['HGT_species_1'].apply(lambda x: x.split(' ')[0])
unique_df['HGT_genus_2'] = unique_df['HGT_species_2'].apply(lambda x: x.split(' ')[0])
```

```python
## Note: there are some MGEs without the information of their position (ISLAND_ND) -> affect the calculation of the statistics
## ISLAND_ND could indicate site-specific recombinase

# how many recombinase clusters are there?
print("Number of recombinase clusters: ", unique_df.recombinase_cluster.nunique())

# how many recombinase types are there?
print("Number of recombinase types: ", unique_df.recombinase.nunique())

# how many MGE types are there?
mge_hgt_type = set(unique_df['recombinase_MGE_assignment_1']).union(set(unique_df['recombinase_MGE_assignment_2']))
print("Number of MGE types: ", len(mge_hgt_type))

# how many species and genera are there?
hgt_species = set(unique_df['HGT_species_1']).union(set(unique_df['HGT_species_2']))
print("Number of species: ", len(hgt_species))

hgt_genus = set(unique_df['HGT_genus_1']).union(set(unique_df['HGT_genus_2']))
print("Number of genus: ", len(hgt_genus))

#how many unique accession ids in proMGE HGT dataset?
hgt_genomes_1 = unique_df[['HGT_genome_1','MGE_position_1']].drop_duplicates()
hgt_genomes_1.rename(columns={"HGT_genome_1":"HGT_genome","MGE_position_1":"MGE_position"}, inplace=True)
hgt_genomes_2 = unique_df[['HGT_genome_2','MGE_position_2']].drop_duplicates()
hgt_genomes_2.rename(columns={"HGT_genome_2":"HGT_genome","MGE_position_2":"MGE_position"}, inplace=True)
hgt_genomes = pd.concat([hgt_genomes_1, hgt_genomes_2]).drop_duplicates()
#hgt_accession_ids = set(unique_df['HGT_genome_1']).union(set(unique_df['HGT_genome_2']))
print("Number of genomes involved in HGT all: ", hgt_genomes['HGT_genome'].nunique())
hgt_genomes_nd = hgt_genomes[hgt_genomes['MGE_position']=="ISLAND_ND"]
print("Number of genomes involved in HGT with ISlAND_ND label: ", hgt_genomes_nd["HGT_genome"].nunique())

#how many HGT events in proMGE HGT dataset?
hgt_events = unique_df[['HGT_genome_1','HGT_genome_2']].drop_duplicates()
print("Number of HGT events all: ", len(hgt_events))

hgt_events_w_nd = unique_df[(unique_df['recombinase_MGE_assignment_1']=='MGE_ND')|(unique_df['recombinase_MGE_assignment_2']=='MGE_ND')]
print("Number of HGT events including MGE_ND: ", len(hgt_events_w_nd))


n_mge_hgt = len(set(unique_df['MGE_position_1']).union(set(unique_df['MGE_position_2'])))
perc_hgt_no_nd = ((n_mge_hgt-1)/promge_df.mge_genome_position.nunique())*100
print(f"Number of MGEs involved in HGT (excluding MGE_ND): {n_mge_hgt-1} ({perc_hgt_no_nd.__round__(3)}%)")

perc_hgt = (len(hgt_genomes)/promge_df.mge_genome_position.nunique())*100
print(f"Number of MGEs involved in HGT: {len(hgt_genomes)} ({perc_hgt.__round__(3)}%)")
```

## Download FASTA files for genomes involved in HGT

```python
# Download FASTA files for genomes involved in HGT according to proMGE database
from Bio import Entrez
from tqdm import tqdm 
import os

Entrez.email = "your.email@example.com"

# genomes involved in HGT
hgt_genomes = hgt_genomes[hgt_genomes['MGE_position']!="ISLAND_ND"]
hgt_genomes['accession'] = hgt_genomes['MGE_position'].apply(lambda x: x.split('.')[-1].split(':')[0])
hgt_accession_ids = hgt_genomes.accession.unique()

fasta_list = []

for acc_id in tqdm(hgt_accession_ids):
    dest_file = f"dataset/genomes/promge/{acc_id}.fasta"

    if not os.path.exists(dest_file):
        try:
            with Entrez.efetch(db="nuccore",
                            id=acc_id,
                            rettype="fasta",
                            retmode="text") as handle:
                fasta_data = handle.read()
                handle.close()
            with open(dest_file, "w") as out:
                out.write(fasta_data)

            #fasta_list.append(fasta_data)
        except Exception as e:
            print(f"Error fetching {acc_id}: {e}")
```

## Create proMGE test data set
- negative sampling on genomes involved in HGT
- most MGEs are located in contigs
- take all regions identified as MGEs (regardless of getting involved in HGT) in each genomes, and treat them as "positives"
- then, perform negative sampling on the remaining regions

### Create promge test data set by using core MGEs defined in the database
- core MGEs contain recombinase families of Xer, Cyan, and Arch1
- additional filter: MGEs with mge_category=Cellular, phage_genes=0, and conjugation_genes=0

```python
# get mges involved in HGT
mge_hgt_df = promge_df[promge_df['mge_genome_position'].isin(hgt_genomes.MGE_position.unique())]

# sample some mges not involved in HGT and from species involved in HGT
mge_core = promge_df[(promge_df['mge_category']=='Cellular') & (promge_df['phage_genes']==0) \
       & (promge_df['conjugation_genes']==0) & (promge_df['recombinase'].isin(['Xer','Cyan','Arch1']))]

# get all biosample ids involved in HGT and retrieve the core islands 
# each biosample contains contigs of the same
mge_hgt_biosample_ids = mge_hgt_df.biosample_id.unique()
promge_neg = mge_core[mge_core['biosample_id'].isin(mge_hgt_biosample_ids)][['Accession','Start','End']]
promge_neg['label'] = 0
promge_neg['prediction_method'] = 'promge'
```

### Read training data pool to determine the taxonomic distance of samples in promge data set

```python
# read who_rki data
who_rki_species = pd.read_excel('dataset/prev_studies/who_rki_sampled_species.xlsx')
```

```python
def define_test_case(df, training_pool):
    training_species = training_pool.species.unique()
    training_genus = training_pool.genus.unique()
    training_family = training_pool.family.unique()
    training_phylum = training_pool.phylum.unique()

    if df['species'] in training_species:
        return "in_species"
    elif df['species'] not in training_species and df['genus'] in training_genus:
        return "out_species"
    elif df['genus'] not in training_genus and df['family'] in training_family:
        return "out_genera_in_families"
    elif df['genus'] not in training_genus and df['family'] not in training_family and df['phylum'] in training_phylum:
        return "out_genera_out_families"
    else:
        return "out_phylum"
```

```python
import taxoniq

# map the taxon to the islandviewer_gis_stats DataFrame
def map_taxon_to_df(row):
    try:
        taxon = taxoniq.Taxon(row['taxid'])
        row['scientific_name'] = taxon.scientific_name

        for t in taxon.ranked_lineage:
            row[t.rank.name] = t.scientific_name 
    except Exception as e:
        row['scientific_name'] = None
    return row

promge_tax_df = mge_hgt_df['taxid'].drop_duplicates().reset_index()

# Apply the mapping function in parallel
promge_tax_info_df = promge_tax_df.parallel_apply(map_taxon_to_df, axis=1)
promge_tax_info_df['test_case'] = promge_tax_info_df.parallel_apply(lambda x: define_test_case(x, who_rki_species), axis=1)
```

```python
tax_test_case_dict = promge_tax_info_df.set_index('taxid')['test_case'].to_dict()
acc_taxid_dict = mge_hgt_df.set_index('Accession')['taxid'].to_dict()
```

```python
promge_test_df_w_info = pd.merge(promge_dataset, mge_hgt_df[['Accession','phylum','family','genus','species']], 
         on='Accession', how='left').drop_duplicates()
```

```python
promge_test_df_w_info['taxid'] = promge_test_df_w_info['Accession'].map(acc_taxid_dict)
promge_test_df_w_info['test_case'] = promge_test_df_w_info['taxid'].map(tax_test_case_dict)
```
